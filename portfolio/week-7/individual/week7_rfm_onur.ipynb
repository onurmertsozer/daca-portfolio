```python
import pandas as pd
import plotly.express as px

# Absolute paths to raw datasets
sales_path = (
    "/Users/onurmertsozer/Desktop/Data Analyst/Week 1 materials-20260824/sales.csv"
)
customers_path = "/Users/onurmertsozer/Desktop/Data Analyst/Week 1 materials-20260824/customers.csv"

# Load sales and customer datasets
df_sales = pd.read_csv(sales_path)
df_customers = pd.read_csv(customers_path)

# Merge sales data with customer details using customer_id
df = pd.merge(df_sales, df_customers, on="customer_id", how="left")

print("✅ Data loaded successfully! Shape:", df.shape)
df.head()
```

```python
# ==========================================
# Task: Role B - Data Cleaning & Validation
# ==========================================

print("=== INITIAL SHAPE ===")
print(df.shape)

# 1. Check and drop duplicate invoices (dedup by invoice_id)
duplicate_count = df.duplicated(subset=['invoice_id']).sum()
print(f"\nDuplicates found: {duplicate_count}")
df = df.drop_duplicates(subset=['invoice_id'], keep='first')

# 2. Check NULLs and drop rows missing critical keys
print("\nNULL counts before cleaning:")
print(df.isnull().sum())
df = df.dropna(subset=['customer_id', 'sale_date', 'total_price'])

# 3. Parse date string to proper datetime format (coerce invalid values to NaT)
df['sale_date'] = pd.to_datetime(df['sale_date'], errors='coerce')
df = df.dropna(subset=['sale_date'])

# 4. Check and filter out invalid/negative amounts
negative_orders = (df['total_price'] <= 0).sum()
print(f"\nNegative/Zero orders removed: {negative_orders}")
df = df[df['total_price'] > 0]

# 5. Cleaning Report Summary
print("\n=== CLEANING REPORT ===")
print(f"Final Cleaned Shape: {df.shape[0]} rows, {df.shape[1]} columns")
print(f"Unique Customers   : {df['customer_id'].nunique()}")
print(f"Date Range         : {df['sale_date'].min()} to {df['sale_date'].max()}")
```
